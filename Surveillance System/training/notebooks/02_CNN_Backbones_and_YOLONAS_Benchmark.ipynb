## 3. Visualization of Training Time vs Accuracy (mAP@0.5)

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x="Model", y="Training Time (s)", palette="viridis")
plt.title("Training Duration (Seconds) across 25 Epochs")
plt.xticks(rotation=30)
plt.ylabel("Seconds")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.scatterplot(data=df, x="FPS", y="mAP@0.5", hue="Model", s=200)
plt.title("Throughput (FPS) vs Detection mAP@0.5")
plt.xlabel("Inference FPS")
plt.ylabel("mAP @ 0.5")
plt.grid(True)
plt.tight_layout()
plt.show()
```
